## Load jax with its settings and import KDExpress

```python
import jax
jax.config.update("jax_enable_x64", True)
jax.config.update("jax_platforms", 'cpu') # put 'gpu' here if you have one
import jax.numpy as jnp
import matplotlib.pyplot as plt
import os,sys
parent_dir = os.getcwd() + '/../src/'
sys.path.append(parent_dir)
from KDExpress import fft_kde1d, binned_kde1d, silverman_bw1d, build_hist_edges
```

 ## Generate random weighted 1-D data 

```python
key = jax.random.PRNGKey(42)
data_1d = jax.random.normal(key, shape=(10_000,))
weights = data_1d**2
plt.hist(data_1d, weights=weights, bins=50, density=True, color='blue', alpha=0.35, label='Data Hist.')
plt.ylabel("PDF")
plt.xlabel("x")
plt.legend()
plt.show()
```

## Define the points where evaluate the PDF and compute the bin edges surrounding them. The histogram done with these bin edges is not very smooth.

```python
points = jnp.linspace(-4, 4, 500)
bin_edges = build_hist_edges(points)
print("Bin edges shape = ", bin_edges.shape)
plt.hist(data_1d, weights=weights, bins=bin_edges, density=True, color='blue', alpha=0.35, label='Data Hist.')
plt.scatter(points, jnp.zeros_like(points)+0.001, c='red', s=5 , label='Points')
plt.ylabel("PDF")
plt.xlabel("x")
plt.legend()
plt.show()
```

## We now compute the KDE as the convolution of the previous wavy histogram with a Gaussian Kernel. The convolution is (efficiently) done using a FFT.

```python
bw = silverman_bw1d(data_1d) # or a user defined value
print("Estimate bandwidth = ", bw)
pdf_est_fft = fft_kde1d(points, data_1d, weights=weights, bin_edges=bin_edges, bw=bw) # here we re-use bin_edges, but it isn't necessary

# jax built in kde
pdf_est_jax = jax.scipy.stats.gaussian_kde(data_1d, weights=weights, bw_method=bw)(points)

# plot
plt.hist(data_1d, weights=weights, bins=bin_edges, density=True, color='blue', alpha=0.35, label='Data Hist.')
plt.scatter(points, jnp.zeros_like(points)+0.001, c='red', s=5 , label='Points')
plt.plot(points, pdf_est_fft, c='black', label='FFT KDE')
plt.plot(points, pdf_est_jax, c='gray', label='Jax built-in KDE')

plt.ylabel("PDF")
plt.xlabel("x")
plt.legend()
plt.show()
```

## Alternatively, we can use the binned KDE. 

```python
pdf_est_binned = binned_kde1d(points, data_1d, weights=weights, bw = bw, nbins=200)

# plot
plt.hist(data_1d, weights=weights, bins=bin_edges, density=True, color='blue', alpha=0.35, label='Data Hist.')
plt.scatter(points, jnp.zeros_like(points)+0.001, c='red', s=5 , label='Points')
plt.plot(points, pdf_est_fft, c='black', label='FFT KDE')
plt.plot(points, pdf_est_jax, c='gray', label='Jax built-in KDE')
plt.plot(points, pdf_est_binned, c='purple', ls='--', label='Binned KDE')

plt.ylabel("PDF")
plt.xlabel("x")
plt.legend()
plt.show()
```
